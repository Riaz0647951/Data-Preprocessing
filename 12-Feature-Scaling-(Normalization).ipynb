```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv(r'.\Datasets\Loan.csv')
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
dataset.describe()
```

```python
sns.distplot(dataset['CoapplicantIncome'])
plt.show()
```

Feature Scaling Technique - Normalization

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
ns = MinMaxScaler()
```

```python
ns.fit(dataset[['CoapplicantIncome']])
```

```python
dataset['CoapplicantIncome_mms'] = ns.transform(dataset[['CoapplicantIncome']])
```

```python
dataset.head(3)
```

```python
# Check graphically whether the nature of data is changed or not
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.title('Before Normalization')
sns.distplot(dataset['CoapplicantIncome'])

plt.subplot(1,2,2)
plt.title('After Normalization')
sns.distplot(dataset['CoapplicantIncome_mms'])
plt.show()
```
