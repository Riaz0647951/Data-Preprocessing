Feature Scaling Technique - Standardization

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
```

```python
ss.fit(dataset[['ApplicantIncome']])
```

```python
dataset['ApplicantIncome_ss'] = pd.DataFrame(ss.transform(dataset[['ApplicantIncome']]), columns=['x'])
```

```python
dataset.head(3)
```

```python
dataset.describe()
```

```python
# check whether the nature of data is changed after standardization or not
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.title('Before Standardization')
sns.distplot(dataset['ApplicantIncome'])

plt.subplot(1,2,2)
plt.title('After Standardization')
sns.distplot(dataset['ApplicantIncome_ss'])
plt.show()
```
