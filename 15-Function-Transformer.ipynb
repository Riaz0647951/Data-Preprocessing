```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
dataset = pd.read_csv(r'.\Datasets\Loan.csv')
dataset.head(3)
```

```python
dataset.isnull().sum()
```

### Function Transformer - Convert Data from Non-Normal Distribution to Normal Distribution

```python
# Check distribution of data
sns.distplot(dataset['CoapplicantIncome'])
plt.show()
```

 - Remove Outliers using IQR Method

```python
Q1 = dataset['CoapplicantIncome'].quantile(0.25)
Q3 = dataset['CoapplicantIncome'].quantile(0.75)
IQR = Q3 - Q1
```

```python
min_range = Q1 - (1.5*IQR)
max_range = Q3 + (1.5*IQR)
min_range, max_range
```

```python
dataset = dataset[dataset['CoapplicantIncome']<=max_range]
```

```python
# Again Check distribution of data
sns.distplot(dataset['CoapplicantIncome'])
plt.show()
```

- Function Transformation

```python
from sklearn.preprocessing import FunctionTransformer
```

```python
# Function Transformer using log1p function
ft = FunctionTransformer(func=np.log1p)
```

```python
ft.fit(dataset[['CoapplicantIncome']])
```

```python
dataset['CoapplicantIncome_ft'] = ft.transform(dataset[['CoapplicantIncome']])
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
sns.distplot(dataset['CoapplicantIncome'])
plt.title('Before Function Transformation')

plt.subplot(1,2,2)
sns.distplot(dataset['CoapplicantIncome_ft'])
plt.title('After Function Transformation')
plt.show()
```

```python
# Function Transformer using x^2 function
ft = FunctionTransformer(func=lambda x: x**2)
ft.fit(dataset[['CoapplicantIncome']])
dataset['CoapplicantIncome_ft1'] = ft.transform(dataset[['CoapplicantIncome']])
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
sns.distplot(dataset['CoapplicantIncome'])
plt.title('Before Function Transformation')

plt.subplot(1,2,2)
sns.distplot(dataset['CoapplicantIncome_ft1'])
plt.title('After Function Transformation')
plt.show()
```
