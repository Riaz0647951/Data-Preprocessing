```python
import pandas as pd
```

```python
dataset = pd.read_csv(r".\Datasets\Loan.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
# Lets encode 'Gender' and 'Married' columns, so first fill missing values in these columns
dataset['Gender'].fillna(dataset['Gender'].mode()[0], inplace=True)
```

```python
dataset['Married'].fillna(dataset['Married'].mode()[0], inplace=True)
```

```python
dataset.isnull().sum()
```

```python
en_data = dataset[['Gender', 'Married']]
```

```python
# One Hot Encoding using pd.get_dummies() but it does not gives the result in 0 and 1 form
pd.get_dummies(en_data)
```

```python
# One Hot Encoding using scikit learn
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
ar = ohe.fit_transform(en_data).toarray()
```

```python
pd.DataFrame(ar, columns=['Gender_Female', 'Gender_Male', 'Married_No', 'Married_Yes'])
```

```python
# As seen above, One Hot Encoding generated 4 columns from the original 2 columns, so the size of data is increased.
# Incase of large dataset, we have to reduce the size by dropping the first column from each category.
new_ohe = OneHotEncoder(drop='first')
new_ar = new_ohe.fit_transform(en_data).toarray()
```

```python
pd.DataFrame(new_ar, columns=['Gender_Male', 'Married_Yes'])
```
